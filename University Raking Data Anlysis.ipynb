```python
numeric_cols = ['scores_teaching', 'scores_research', 'scores_citations',
                'scores_industry_income', 'scores_international_outlook',
                'stats_student_staff_ratio']
rank[numeric_cols] = rank[numeric_cols].fillna(rank[numeric_cols].mean())

# For object columns (like strings), fill with a placeholder (e.g., "Unknown")
object_cols = ['stats_female_male_ratio', 'subjects_offered']
rank[object_cols] = rank[object_cols].fillna('Unknown')

# If overall_score is object but represents numeric scores, try converting and filling
rank['overall_score'] = pd.to_numeric(rank['overall_score'], errors='coerce')
rank['overall_score'] = rank['overall_score'].fillna(rank['overall_score'].mean())
```

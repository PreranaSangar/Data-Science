# Chi-square Test

```python
# Chi-Square Test for Independence
# Problem: Association between Device Type and Customer Satisfaction
```

```python
import pandas as pd
from scipy.stats import chi2_contingency, chi2
```

```python
# Step 1: Create Contingency Table (Observed Frequencies)
# --------------------------------------------------------
# Data from the problem statement
data = {
    "Very Satisfied": [50, 70],
    "Satisfied": [80, 100],
    "Neutral": [60, 90],
    "Unsatisfied": [30, 50],
    "Very Unsatisfied": [20, 50]
}

# Index represents device type: Smart Thermostat vs Smart Light
contingency_table = pd.DataFrame(data, index=["Smart Thermostat", "Smart Light"])

print("Contingency Table (Observed Frequencies):\n")
print(contingency_table, "\n")
```

```python
# Step 2: Define Hypotheses
# --------------------------
# Null Hypothesis (H0): There is no significant association between device type and customer satisfaction.
# Alternative Hypothesis (H1): There is a significant association between device type and customer satisfaction.
```

```python
# Step 3: Perform Chi-Square Test
# --------------------------------
chi2_stat, p_val, dof, expected = chi2_contingency(contingency_table)
```

```python
# Step 4: Calculate Critical Value
# ---------------------------------
alpha = 0.05  # significance level
critical_value = chi2.ppf(q=1-alpha, df=dof)
```

```python
# Step 5: Print Results
# ----------------------
print("Chi-Square Test Results:\n")
print(f"Chi-Square Statistic = {chi2_stat:.4f}")
print(f"Degrees of Freedom = {dof}")
print(f"Critical Value (alpha=0.05) = {critical_value:.4f}")
print(f"p-value = {p_val:.4f}\n")

print("Expected Frequencies (calculated under H0):\n")
print(pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns), "\n")
```

```python
# Step 6: Decision
# -----------------
if chi2_stat > critical_value:
    print("Decision: Reject Null Hypothesis (There is a significant association).")
else:
    print("Decision: Fail to Reject Null Hypothesis (No significant association).")
```

# Hypothesisi Testing

```python
# Hypothesis Testing for Weekly Operating Costs
# Problem: Investigating if the actual weekly operating costs are higher than the model suggests.
```

```python
import math
from scipy.stats import norm
```

```python
# Step 1: State the Hypotheses

# Null Hypothesis (H0): The mean weekly operating cost = Theoretical cost (no increase).
# Alternative Hypothesis (H1): The mean weekly operating cost > Theoretical cost (increase).
```

```python
# Step 2: Define Given Data

sample_mean = 3050        # x̄ = sample mean weekly cost
sample_size = 25          # n = number of restaurants
X_mean = 600              # mean units produced
X_std = 25                # standard deviation of units produced

# Theoretical model: W = 1000 + 5X
theoretical_mean = 1000 + 5 * X_mean   # μ = theoretical mean cost
sigma = 5 * X_std                      # σ = standard deviation of weekly cost
```

```python
# Step 3: Calculate Test Statistic (Z-test since σ is known)
# Formula: Z = (x̄ - μ) / (σ / √n)

Z = (sample_mean - theoretical_mean) / (sigma / math.sqrt(sample_size))
```

```python
# Step 4: Determine Critical Value at alpha = 0.05 (one-tailed test)
alpha = 0.05
critical_value = norm.ppf(1 - alpha)   # 95th percentile for one-tailed test
```

```python
# Step 5: Make Decision

if Z > critical_value:
    decision = "Reject Null Hypothesis (Costs are significantly higher)."
else:
    decision = "Fail to Reject Null Hypothesis (No significant evidence of higher costs)."
```

```python
# Step 6: Print Results

print("Hypothesis Testing Report\n")
print("Step 1: Hypotheses")
print("H0: Mean weekly cost = Theoretical cost")
print("H1: Mean weekly cost > Theoretical cost\n")

print("Step 2: Data Summary")
print(f"Sample Mean (x̄): {sample_mean}")
print(f"Theoretical Mean (μ): {theoretical_mean}")
print(f"Standard Deviation (σ): {sigma}")
print(f"Sample Size (n): {sample_size}\n")

print("Step 3: Test Statistic")
print(f"Z = {Z:.4f}\n")

print("Step 4: Critical Value")
print(f"Critical Value (α=0.05): {critical_value:.4f}\n")

print("Step 5: Decision")
print(decision)
```

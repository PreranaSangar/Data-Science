# Z Score

```python
#Z test is astatistical test to determine wheter
#two population means are different when the variance are know
#sample size is large
#means of two group
```

```python
import numpy as np
import scipy.stats as stats
```

```python
sample_mean=80
population_mean=100
sample_std=15
sample_size=50
alpha=0.05
```

```python
import numpy as np
z_score=(sample_mean-population_mean)/(sample_std/np.sqrt(50))
```

```python
z_score
```

```python
z_critical=stats.norm.ppf(1-alpha) 
# ppf : stands for percent point function it is the inverse of cumulative distribution function
#norm= normally distributes curve
```

```python
#from sklearn.preprocessing import StandardScaler
```

```python
z_critical
```

```python
if z_score>z_critical:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

```python
p_value=1-stats.norm.cdf(z_score)
# cdf= Area under curve
```

```python
# Bar Chart
import matplotlib.pyplot as plt

x = ['Prerana', 'Sanwi', 'Onkar', 'Rahul', 'Abhikesh', 'Arya']
y = [100, 243, 364, 140, 90, 300]
z = 'hotpink', 'green', 'blue', 'yellow', 'c', 'darkorange'

plt.title('Student vs Student Subject Marks',fontsize=20)
plt.xlabel('Students',fontsize=10)
plt.ylabel('Student Marks',fontsize=10)

# One label for the bar chart
plt.bar(x, y, color=z, label=x,edgecolor='black',width=0.7)
plt.legend()
plt.show()
```

```python
x=['oppo','Realme','Moto','Redme','samsung']
y=[10,8,18,12,9]
z='m','g','c','r','k'
plt.xlabel('Price of Mobile in K',fontsize=10)
plt.ylabel('Mobile',fontsize=10)
plt.title('Mobile vs Mobile Price',fontsize=20)
plt.barh(x,y,color=z,label=x,height=0.6,edgecolor='black')
plt.legend()
plt.show()
```

```python
# Pie Chart
x=['oppo','Realme','Moto','Redme','samsung']
y=[10,8,18,12,9]
z='m','g','c','r','y'
s=[0,0,0.07,0,0]
plt.title('Mobile vs Mobile Price',fontsize=20)
plt.pie(y,labels=x,autopct='%2.2f%%',shadow='True',explode=s,colors=z) # 2 is always fix
plt.legend()
plt.show()
```

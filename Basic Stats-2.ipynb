```python
import numpy as np
import scipy.stats as stats
```

```python
# Step 1: Input Data
# -----------------------------
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32,
                 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
n = len(data)                     # Sample size
mean = np.mean(data)               # Sample mean
sample_std = np.std(data, ddof=1)  # Sample standard deviation (ddof=1 for sample formula)

# Significance level for 99% confidence
alpha = 0.01
```

```python
# (a) Using SAMPLE standard deviation → t-distribution
```

```python
# Step 2: Find t-critical value from the t-distribution
# df = n - 1 degrees of freedom
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)
```

```python
# Step 3: Compute margin of error using t-distribution
margin_of_error_t = t_critical * (sample_std / np.sqrt(n))
```

```python
# Step 4: Construct confidence interval
ci_t = (mean - margin_of_error_t, mean + margin_of_error_t)
```

```python
# (b) Using KNOWN population standard deviation → z-distribution
```

```python
pop_std = 0.2  # Given population standard deviation
```

```python
# Step 5: Find z-critical value from the standard normal distribution
z_critical = stats.norm.ppf(1 - alpha/2)
```

```python
# Step 6: Compute margin of error using z-distribution
margin_of_error_z = z_critical * (pop_std / np.sqrt(n))
```

```python
# Step 7: Construct confidence interval
ci_z = (mean - margin_of_error_z, mean + margin_of_error_z)
```

```python
# Step 8: Display Results
print("99% Confidence Interval using sample std (t-distribution):", ci_t)
print("99% Confidence Interval using population std (z-distribution):", ci_z)
```

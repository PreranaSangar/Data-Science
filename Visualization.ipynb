```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
from sklearn.datasets import load_iris
```

```python
# For reproducibility
np.random.seed(42)
```

```python
# 1) Line plot — Time series
dates = pd.date_range(start="2025-01-01", periods=100, freq="D")
dates
```

```python
sales = np.cumsum(np.random.randn(100) * 5 + 50)
sales
```

```python
ts_df = pd.DataFrame({"date": dates, "sales": sales}).set_index("date")
```

```python
ts_df
```

```python
plt.figure(figsize=(10,4))
plt.plot(ts_df.index, ts_df["sales"])
plt.title("Line plot — Time series (Sales)")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
# 2) Bar chart — Category counts
categories = np.random.choice(["A", "B", "C", "D"], size=200, p=[0.2,0.35,0.3,0.15])
categories
```

```python
cat_counts = pd.Series(categories).value_counts().sort_index()
cat_counts
```

```python
plt.figure(figsize=(6,4))
plt.bar(cat_counts.index, cat_counts.values)
plt.title("Bar chart — Category counts")
plt.xlabel("Category")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# 3) Histogram — Distribution
vals = np.random.normal(loc=0, scale=1.5, size=1000)
vals
```

```python
plt.figure(figsize=(6,4))
plt.hist(vals, bins=30)
plt.title("Histogram — Distribution of values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# 4) Scatter plot with Trendline
x = np.random.rand(200) * 10
y = 2.5 * x + np.random.randn(200) * 4 + 5
```

```python
x
```

```python
y
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x, y)
# Add trendline
m, b = np.polyfit(x, y, 1)
plt.plot(np.sort(x), m * np.sort(x) + b, color="red")
plt.title("Scatter plot — X vs Y with trendline")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.show()
```

```python
# 5) Boxplot — Compare distributions
grouped_vals = [np.random.normal(loc=i, scale=1.0, size=100) for i in range(4)]
plt.figure(figsize=(6,4))
plt.boxplot(grouped_vals, labels=["G1","G2","G3","G4"])
plt.title("Boxplot — Group-wise distribution comparison")
plt.ylabel("Value")
plt.tight_layout()
plt.show()
```

```python
# 6) Heatmap — Correlation matrix
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
corr = iris_df.corr()
```

```python
plt.figure(figsize=(6,5))
plt.imshow(corr, vmin=-1, vmax=1, aspect='auto')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha='right')
plt.yticks(range(len(corr.index)), corr.index)
plt.title("Heatmap — Correlation matrix (Iris features)")
plt.tight_layout()
plt.show()
```

```python
# 7) Pie chart — Category proportions
plt.figure(figsize=(6,6))
plt.pie(cat_counts.values, labels=cat_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Pie chart — Category proportions")
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
plt.rcParams['figure.figsize']=8,3
sns.set_theme(style="whitegrid", palette="dark")
```

```python
personality=pd.read_csv(r'C:\Users\admin\Downloads\archive\personality_dataset.csv')
```

```python
personality.isnull().sum()
```

```python

numerical_cols = ['Time_spent_Alone', 'Social_event_attendance', 'Going_outside', 'Friends_circle_size', 'Post_frequency']
personality[numerical_cols] = personality[numerical_cols].fillna(personality[numerical_cols].mean())

# Fill categorical columns with mode (most frequent value)
categorical_cols = ['Stage_fear', 'Drained_after_socializing']
for col in categorical_cols:
    personality[col] = personality[col].fillna(personality[col].mode()[0])
```
